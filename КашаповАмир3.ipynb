```python
#Кашапов Амир
#1
stroka = input()
spisok = input().split()
if stroka in spisok:
  print(spisok.index(stroka))
else:
  print('Такой нет')
```

```python
#2
spisok = input().split()
print(list(set(spisok)))
```

```python
#3
spisok = input().split()
slovar = {}
for char in spisok:
  if char in slovar:
    slovar[char] += 1
  else:
    slovar[char] = 1
print(slovar)
```

```python
spisok = [int(x) for x in input().split()]
spisok.remove(min(spisok))
print(spisok)
```
