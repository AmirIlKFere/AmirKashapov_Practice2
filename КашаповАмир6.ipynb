```python
#Кашапов Амир 36-64
class Student:
    def __init__(self, name, group, tel_number):
        self.name = name
        self.group = group
        self.grades = []
        self.__tel_number = tel_number

    def add_grade(self, grade):
        self.grades.append(grade)

    def get_average(self):
        if not self.grades:
            return 0
        return sum(self.grades) / len(self.grades)

    def info(self):
        print(f"Имя: {self.name}")
        print(f"Группа: {self.group}")
        print(f"Оценки: {self.grades}")
        print(f"Средний балл: {self.get_average():.2f}")
        print(f"Телефон: {self.__tel_number}")
        print("-" * 30)


kamina = Student("Камина", "Механика-1", "+7-900-111-22-33")
kamina.add_grade(3)
kamina.add_grade(4)
kamina.add_grade(2)
kamina.add_grade(5)
kamina.add_grade(3)

simon = Student("Симон", "Механика-1", "+7-900-444-55-66")
simon.add_grade(4)
simon.add_grade(4)
simon.add_grade(3)
simon.add_grade(5)
simon.add_grade(4)

yoko = Student("Ёко", "Механика-1", "+7-900-777-88-99")
yoko.add_grade(5)
yoko.add_grade(5)
yoko.add_grade(4)
yoko.add_grade(5)
yoko.add_grade(5)

kamina.info()
simon.info()
yoko.info()
```
