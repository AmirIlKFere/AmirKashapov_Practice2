```python
n = int(input())

total_age = 0
for i in range(1, n + 1):
    age = int(input(f"Введите возраст {i}-го человека: "))
    total_age += age

average = total_age / n
print(average)
```

```python
code = input().strip().lower()

r = int(code[0:2], 16)
g = int(code[2:4], 16)
b = int(code[4:6], 16)

if code == "000000":
    print("черный")
elif code == "ffffff":
    print("белый")
elif r == g == b:
    print("серый")
else:
    if r >= g and r >= b:
        print("красный")
    elif g >= r and g >= b:
        print("зеленый")
    else:
        print("синий")
```
