```python
#Кашапов Амир
#1
def calculator(num1, num2, oper):
  if oper == '+':
    return(num1+num2)
  elif oper == '-':
    return(num1 - num2)
  elif oper == '*':
    return(num1 * num2)
  elif oper == '/':
    return(num1 / num2)
```

```python
#2
def mesyaci(month):

  if 1 <= month <= 3:
      kvartal = "1 квартал"
  elif 4 <= month <= 6:
      kvartal = "2 квартал"
  elif 7 <= month <= 9:
      kvartal = "3 квартал"
  else:
      kvartal = "4 квартал"

  if month in [12, 1, 2]:
      season = "зима"
  elif 3 <= month <= 5:
      season = "весна"
  elif 6 <= month <= 8:
      season = "лето"
  elif 9 <= month <= 11:
      season = "осень"

  return(kvartal, season)
```

```python
#3
def characters(stroka, chars):
    char_count = {}

    for char in chars:
        char_count[char] = stroka.count(char)

    return char_count
```

```python
#4
def m_z(stroka):
    count = 0
    for char in stroka:
        if 'm' <= char <= 'z':
            count += 1

    return count
```
